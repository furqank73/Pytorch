Here is the complete walkthrough in formatted Markdown text.

---

# Text to Indices: Step-by-Step Example

### 1. The Setup (Your Vocabulary Data)

Before processing any input, define the known word dictionary (`vocab`). Every word points to a specific numerical ID, and unrecognized words fall back to `<UNK>`.

* `<UNK>` $\rightarrow$ `0` (Unknown word placeholder)
* `hello` $\rightarrow$ `1`
* `world` $\rightarrow$ `2`

---

### 2. User Input

The user inputs the sentence:

> **`"hello aliens"`**

---

### 3. Step-by-Step Breakdown

1. **Tokenize the input**
The sentence is split into individual tokens (words):
`["hello", "aliens"]`
2. **Initialize an empty list**
`indexed_text = []`
3. **Process Token 1: `"hello"**`
* **Check:** Is `"hello"` in `vocab`?
* **Result:** **Yes**. Found in vocabulary.
* **Action:** Retrieve `vocab["hello"]`, which equals `1`.
* **State of list:** `indexed_text = [1]`


4. **Process Token 2: `"aliens"**`
* **Check:** Is `"aliens"` in `vocab`?
* **Result:** **No**. Not in vocabulary.
* **Action:** Fall back to the default token `vocab["<UNK>"]`, which equals `0`.
* **State of list:** `indexed_text = [1, 0]`



---

### 4. Summary Table

| Token | Found in `vocab`? | Value Assigned | Current `indexed_text` |
| --- | --- | --- | --- |
| `"hello"` | **Yes** | `1` | `[1]` |
| `"aliens"` | **No** (Defaults to `<UNK>`) | `0` | `[1, 0]` |

---

### 5. Final Result

The function returns the final list of numerical IDs:

`[1, 0]`

```python
def text_to_indices(text,vocab):
    indexed_text=[]
    for token in tokenize(text):
        if token in vocab:
            indexed_text.append(vocab[token])
        else:
            indexed_text.append(vocab["<UNK>"])
    return indexed_text
```

```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
class Qadataset(Dataset):
    def __init__(self,df,vocab):
        self.df=df
        self.vocab=vocab

    def __len__(self):
        return self.df.shape[0]
    
    def __getitem__(self,index):
        numerical_question=text_to_indices(self.df.iloc[index]["question"],self.vocab)
        numerical_answer=text_to_indices(self.df.iloc[index]["answer"],self.vocab)

        return torch.tensor(numerical_question),torch.tensor(numerical_answer)
```

```python
dataset=Qadataset(df,vocab)
```

```python
dataloader=DataLoader(dataset,batch_size=1,shuffle=True)
```

```python
for question,answer in dataloader:
    print(question,answer[0])
```

```python
import torch.nn as nn

class simplernn(nn.Module):
    def __init__(self,vocab_size):
        super().__init__()
        self.embedding=nn.Embedding(vocab_size,embedding_dim=50)
        self.rnn=nn.RNN(50,64,batch_first=True)
        self.fc=nn.Linear(64,vocab_size)

    def forward(self,question):
        embedded_question=self.embedding(question)
        hidden,final=self.rnn(embedded_question)
        output=self.fc(final.squeeze(0))
        return output
```

```python
learning_rate=0.001
epochs=25
```

```python
model=simplernn(len(vocab))
```

```python
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=learning_rate)
```

```python
for epoch in range(epochs):
    total_loss=0
    for question,answer in dataloader:
        optimizer.zero_grad()
        output=model(question)
        loss=criterion(output,answer[:,0])
        loss.backward()
        optimizer.step()
        total_loss=total_loss + loss.item()

    print(f"epochs:{epoch+1},loss :{total_loss:4f}")
```

```python
def predict(model, question, threshold=0.5):

  # convert question to numbers
  numerical_question = text_to_indices(question, vocab)

  # tensor
  question_tensor = torch.tensor(numerical_question).unsqueeze(0)

  # send to model
  output = model(question_tensor)

  # convert logits to probs
  probs = torch.nn.functional.softmax(output, dim=1)

  # find index of max prob
  value, index = torch.max(probs, dim=1)

  if value < threshold:
    print("I don't know")

  print(list(vocab.keys())[index])
```

```python
import torch

def calculate_accuracy(model, data_loader):
    model.eval()  # Put the model in evaluation mode
    correct = 0
    total = 0

    with torch.no_grad():  # Turn off gradient calculation to save memory and speed up
        for questions, answers in data_loader:
            # 1. Forward pass: get model predictions
            outputs = model(questions)  # Shape: [batch_size, vocab_size]
            
            # 2. Pick the word ID with the highest score
            # torch.argmax finds the index of the largest value across columns (dim=1)
            predictions = torch.argmax(outputs, dim=1)
            
            # 3. Handle single-token targets:
            # If answers is 2D (like [batch, 1]), flatten to match predictions [batch]
            targets = answers.squeeze()
            
            # 4. Count matches
            correct += (predictions == targets).sum().item()
            total += targets.size(0)

    # 5. Calculate percentage
    accuracy = (correct / total) * 100
    return accuracy
```
