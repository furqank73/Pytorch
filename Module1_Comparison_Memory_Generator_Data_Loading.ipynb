# Module 1: Comparison of Memory-Based and Generator-Based Data Loading

This notebook completes the required tasks for comparing data loading approaches.
It loads image paths, inspects image dimensions, displays samples, and counts agricultural images.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

base_dir = './images_dataSAT'
if not os.path.exists(base_dir):
    alt_dir = os.path.join('.', 'AI Capstone DL Projects', 'CNN Model Development', 'images_dataSAT')
    if os.path.exists(alt_dir):
        base_dir = alt_dir

print('Dataset directory:', base_dir)

dir_non_agri = os.path.join(base_dir, 'class_0_non_agri')
dir_agri = os.path.join(base_dir, 'class_1_agri')

print('Non-agri folder exists:', os.path.exists(dir_non_agri))
print('Agri folder exists:', os.path.exists(dir_agri))
```

## Task 1: Determine the shape (dimensions) of a single image stored in the image_data variable.

We open the first image in the non-agricultural folder and store it in `image_data`.

```python
non_agri_files = sorted(os.listdir(dir_non_agri))
first_non_agri_path = os.path.join(dir_non_agri, non_agri_files[0])
image_data = np.array(Image.open(first_non_agri_path))

print('Image shape:', image_data.shape)
```

## Task 2: Display the first four images in the './images_dataSAT/class_0_non_agri/' directory.

The loop below loads and shows four sample non-agricultural images.

```python
non_agri_paths = [os.path.join(dir_non_agri, filename) for filename in sorted(os.listdir(dir_non_agri))]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, path in zip(axes, non_agri_paths[:4]):
    img = Image.open(path)
    ax.imshow(img)
    ax.axis('off')
    ax.set_title('Non-agri')

plt.tight_layout()
plt.show()
```

## Task 3: Create a list named agri_images_paths that contains the full file paths of all images in dir_agri.

The list is sorted before saving it.

```python
agri_images_paths = sorted(
    os.path.join(dir_agri, filename)
    for filename in os.listdir(dir_agri)
)

print('Number of files in agri_images_paths:', len(agri_images_paths))
print('First 3 entries:', agri_images_paths[:3])
```

## Task 4: Determine the number of agricultural images in './images_dataSAT/class_1_agri/' .

We use the length of the sorted list of paths as the count.

```python
agri_count = len(agri_images_paths)
print('Agricultural image count:', agri_count)
```

## Task 5: Display the first four images of agricultural land.

We show the first four images from the agricultural folder.

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, path in zip(axes, agri_images_paths[:4]):
    img = Image.open(path)
    ax.imshow(img)
    ax.axis('off')
    ax.set_title('Agri')

plt.tight_layout()
plt.show()
```
