# Tokenize The Text

````python
def tokenize(text):
    text=text.lower()
    text=text.replace("?","")
    text=text.replace("'","")
    return text.split() #Split into words (tokens)
````

## When building the vocabulary, we reserve ID 0 for unknown words.
<UNK> stands for Unknown. so if unknown word found in input we want to predict from training it will be unk

````python
vocab = {'<UNK>':0}
````

## build_vocab() takes a question-answer pair, tokenizes the text, and adds every new word to the vocabulary with a unique integer ID so the NLP model can later convert words into numbers.

````python
def build_vocab(row):
    tokenised_question=tokenize(row['question'])
    tokenised_answer=tokenize(row['answer'])

    merged_tokens=tokenised_question + tokenised_answer

    for token in merged_tokens: #for token in merged_tokens:Loops through each word in that combined list.vocab[token] = len(vocab)
        #Assigns a unique integer index to the word based on the current size of vocab.
        # If vocab is empty (len = 0), the first word gets assigned index 0.The next word gets index 1, the next gets 2, and so on.
        if token not in vocab:
            vocab[token] = len(vocab)
````

### Vocabulary

* `<UNK>` $\rightarrow$ `0`
* `hello` $\rightarrow$ `1`
* `world` $\rightarrow$ `2`

---

### Input: `"hello aliens"`

1. **Split into words:** `["hello", "aliens"]`
2. **`"hello"`:** Found in vocab $\rightarrow$ add `1`
3. **`"aliens"`:** Not found $\rightarrow$ add `<UNK>` (`0`)

---

### Result: `[1, 0]`

````python
def text_to_indices(text,vocab):
    indexed_text=[]
    for token in tokenize(text):
        if token in vocab:
            indexed_text.append(vocab[token])
        else:
            indexed_text.append(vocab['<UNK>'])
    return indexed_text
````

````python
import torch
from torch.utils.data import Dataset, DataLoader
````

````python
class qadataset(Dataset):
    def __init__(self,df,vocab):
        self.df=df       #df → your question-answer DataFrame
        self.vocab=vocab #vocab → your word-to-number dictionary

    def __len__(self):
        return self.df.shape[0]
    
    def __getitem__(self,index):
        numerical_question=text_to_indices(self.df.iloc[index]['question'],self.vocab) #Row Retrieval of both: Grabs the row at position index using .iloc[index].
        numerical_answer=text_to_indices(self.df.iloc[index]['answer'],self.vocab)

        return torch.tensor(numerical_question), torch.tensor(numerical_answer)
````

````python
import torch.nn as nn
````

````python
class simplernn(nn.Module):
    def __init__(self,vocab_size):
        super().__init__()
        self.embedding=nn.Embedding(vocab_size,embedding_dim=50)
        self.rnn=nn.RNN(50,64,batch_first=True)
        self.fc=nn.Linear(64,vocab_size)# fc stands for Fully Connected layer.Takes the 64-number summary of the sentence 
                                        # and projects it across the entire vocabulary so each word gets a score.

    def forward(self,question):
        embedded_question=self.embedding(question)#Converts words from plain IDs (like [4, 12, 9]) into lists of 50 numbers each
        hidden,final=self.rnn(embedded_question)
        output=self.fc(final.squeeze(0))

        return output
````

### Concrete Example Walkthrough

Let's say our vocabulary has only **5 words**:

| Word | ID |
| --- | --- |
| `<PAD>` | 0 |
| `how` | 1 |
| `are` | 2 |
| `you` | 3 |
| `good` | 4 |

So, `vocab_size = 5`.

#### Step 1: Input

You send in the question: `"how are you"`

As IDs, this is: `[1, 2, 3]`

#### Step 2: Embedding

* `1` (`how`) $\rightarrow$ list of 50 numbers
* `2` (`are`) $\rightarrow$ list of 50 numbers
* `3` (`you`) $\rightarrow$ list of 50 numbers

#### Step 3: RNN Reads

* Reads `how` $\rightarrow$ updates memory
* Reads `are` $\rightarrow$ updates memory
* Reads `you` $\rightarrow$ finishes reading.
* Result (`final`): **one summary list of 64 numbers**.

#### Step 4: Linear Layer (`fc`)

The layer takes those 64 numbers and outputs 5 scores (one for each word in the vocabulary):

```text
Score for <PAD>: -1.2
Score for how:    0.4
Score for are:   -0.1
Score for you:    0.2
Score for good:   4.8   <-- Highest score!

```

The model picks **`good`** (ID 4) as its predicted answer.

````python
model = simplernn(len(vocab))
````

````python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
````

````python
# training loop

for epoch in range(epochs):

  total_loss = 0

  for question, answer in dataloader:

    optimizer.zero_grad()

    # forward pass
    output = model(question)

    # target: answer sequence
    # output shape: (batch, vocab_size)
    # answer[:, 0] is only the first word, so we use the whole answer sequence
    # for a simpler fix, model should output one logit per token class, but this notebook
    # is set up as a single-token classifier. Keep the existing structure and train on the first answer token.
    loss = criterion(output, answer[:, 0])

    # gradients
    loss.backward()

    # update
    optimizer.step()

    total_loss = total_loss + loss.item()

  print(f"Epoch: {epoch+1}, Loss: {total_loss:4f}")
````

````python
def predict(model, question, threshold=0.5):

  # convert question to numbers
  numerical_question = text_to_indices(question, vocab)

  # tensor
  question_tensor = torch.tensor(numerical_question).unsqueeze(0)

  # send to model
  output = model(question_tensor)

  # convert logits to probs
  probs = torch.nn.functional.softmax(output, dim=1)

  # find index of max prob
  value, index = torch.max(probs, dim=1)

  if value < threshold:
    print("I don't know")

  print(list(vocab.keys())[index])
````

````python
# Assuming you have a test_loader or train_loader:
# Use the existing DataLoader variable: dataloader
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for questions, answers in dataloader:
        predictions = model(questions).argmax(dim=1)
        targets = answers[:, 0]

        correct += (predictions == targets).sum().item()
        total += targets.numel()

acc = 100 * correct / total
print(f"Model Accuracy: {acc:.2f}%")
print(f"Model Accuracy: {acc:.2f}%")
````

### RNN-Based Question Answering System — Summary

#### Goal

Build a simple NLP model that can answer a question by predicting a single-word answer.

Example:

**Question:**

> What is the capital of France?

**Answer:**

> Paris

---

### Workflow

#### 1. Load Dataset

The dataset contains question-answer pairs.

| Question                       | Answer  |
| ------------------------------ | ------- |
| What is the capital of France? | Paris   |
| What is the largest planet?    | Jupiter |

---

#### 2. Tokenization

Convert text into words.

```text
What is AI?
```

↓

```python
["what", "is", "ai"]
```

---

#### 3. Build Vocabulary

Assign an ID to every unique word.

```python
{
    "<UNK>":0,
    "what":1,
    "is":2,
    "paris":3
}
```

---

#### 4. Convert Text to Numbers

Neural networks cannot understand words.

```text
what is france
```

↓

```python
[1, 2, 5]
```

---

#### 5. Create Dataset & DataLoader

Prepare question-answer pairs for training.

```python
(question_tensor, answer_tensor)
```

---

#### 6. Embedding Layer

Convert word IDs into dense vectors.

```python
1
```

↓

```python
[0.12, -0.45, 0.88, ...]
```

---

#### 7. RNN Layer

Reads words one by one and remembers context.

```text
what → is → the → capital → of → france
```

The final hidden state represents the meaning of the whole question.

---

#### 8. Linear Layer

Uses the RNN output to predict the answer word.

```text
Question
   ↓
Embedding
   ↓
RNN
   ↓
Linear
   ↓
Predicted Answer
```

---

#### 9. Training

For each question:

1. Forward pass
2. Calculate loss
3. Backpropagation
4. Update weights

The model gradually learns the correct answers.

---

#### 10. Prediction

When a new question is given:

```text
What is the largest planet in our solar system?
```

↓

```text
Jupiter
```

(if it learned that pattern from training data)

---

### Architecture

```text
Question
   ↓
Tokenization
   ↓
Vocabulary Mapping
   ↓
Embedding Layer
   ↓
RNN Layer
   ↓
Linear Layer
   ↓
Answer Word
```
