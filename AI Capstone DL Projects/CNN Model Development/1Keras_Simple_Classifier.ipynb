# Keras Simple Image Classifier

This notebook shows a simplified version of a Keras image classification workflow.

The goal is to:
- prepare image data,
- build a small CNN model,
- train it on labeled images,
- and evaluate its accuracy.

This version is easier to read than the full lab notebook and focuses on the key ideas only.

## 1. Import libraries

We use TensorFlow and Keras for the model, and `ImageDataGenerator` to load images from folders.

```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print("TensorFlow version:", tf.__version__)
```

## 2. Set dataset path

The folder contains two classes, for example:
- agricultural images
- non-agricultural images

`flow_from_directory` automatically reads the folders and assigns labels.

```python
DATASET_PATH = os.path.join('.', 'images_dataSAT')
print("Dataset path:", DATASET_PATH)
```

## 3. Prepare image data

We normalize image pixels to [0, 1] and split data into training and validation sets.

```python
img_size = (64, 64)
batch_size = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2,
    horizontal_flip=True,
    rotation_range=20,
)

train_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='binary',
    subset='training'
)

validation_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='binary',
    subset='validation'
)

print(train_generator.class_indices)
```

## 4. Build a simple CNN model

This model has a few convolutional layers, pooling layers, and a final dense output layer for binary classification.

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

## 5. Train the model

The model learns patterns from the training images and validates its progress on unseen validation images.

```python
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=5,
    verbose=1
)
```

## 6. Evaluate the model

After training, we check the validation accuracy and the final model quality.

```python
val_loss, val_accuracy = model.evaluate(validation_generator, verbose=0)
print(f"Validation accuracy: {val_accuracy:.4f}")
print(f"Validation loss: {val_loss:.4f}")
```

## Summary

This example shows the basic Keras workflow for an image classifier:
1. load images from folders,
2. build a CNN,
3. train with `fit`,
4. evaluate accuracy.

This is the simplest form of a deep learning image classification notebook.
