# Lab 2: Simple Keras Image Classifier

This notebook contains a simplified Keras version of the image classification task.

The purpose is to make the workflow easier to understand:
- load training images,
- build a small CNN,
- train the model,
- and check validation performance.

This version focuses only on the core concepts without the extra complexity from the full lab notebook.

## 1. Import the required libraries

We need TensorFlow and Keras for model creation and image loading.

```python
import os
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print("TensorFlow version:", tf.__version__)
```

## 2. Set the dataset folder

The dataset is organized into class folders, so `ImageDataGenerator` can read labels automatically.

```python
DATASET_PATH = os.path.join('.', 'images_dataSAT')
print("Dataset path:", DATASET_PATH)
```

## 3. Prepare the data

We normalize pixel values to the range [0, 1] and split the data into training and validation sets.

```python
img_size = (64, 64)
batch_size = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2,
    horizontal_flip=True,
    rotation_range=20
)

train_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='binary',
    subset='training'
)

validation_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='binary',
    subset='validation'
)

print(train_generator.class_indices)
```

## 4. Define the model

This is a small CNN with convolution layers, pooling, and a final output layer.

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

## 5. Train the model

`model.fit()` trains the model automatically using the generated data.

```python
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=3,
    verbose=1
)
```

## 6. Evaluate the model

After training, we test the model on the validation set.

```python
val_loss, val_accuracy = model.evaluate(validation_generator, verbose=0)
print(f'Validation accuracy: {val_accuracy:.4f}')
print(f'Validation loss: {val_loss:.4f}')
```

## Summary

This simplified Keras notebook shows the main workflow for image classification:
1. load the dataset,
2. create a CNN,
3. train the model,
4. evaluate validation accuracy.

Keras is beginner-friendly because the training process is compact and easy to follow.
