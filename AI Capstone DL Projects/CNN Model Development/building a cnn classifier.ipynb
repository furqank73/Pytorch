## 🧠 CNN Classifier — Simple Summary

### 1. Why use CNNs?

**CNNs (Convolutional Neural Networks)** are very effective for **image classification**.

They can learn visual patterns such as:

* Edges
* Shapes
* Textures
* Objects

A CNN usually performs better and is more robust than manually creating rules for recognizing images.

---

### 2. CNN Image Classification Pipeline

The overall process is:

**📁 Collect & label images**
↓
**🔄 Load data efficiently**
↓
**🎨 Apply data augmentation**
↓
**🧠 Build CNN**
↓
**🏋️ Train CNN**
↓
**📊 Evaluate model**

---

## 3. Keras vs PyTorch

| Feature     | Keras             | PyTorch                   |
| ----------- | ----------------- | ------------------------- |
| Level       | High-level        | More detailed/lower-level |
| Code        | Less code         | More code                 |
| Training    | `model.fit()`     | Write training loop       |
| Flexibility | Easier            | More control              |
| Best for    | Quick prototyping | Custom research/debugging |

### Easy way to remember:

**Keras = Simple & fast to build**
**PyTorch = More control & customization**

---

## 4. Data Loading

Both frameworks use **lazy loading**.

### What does lazy loading mean?

Instead of loading the entire dataset into RAM at once:

> Images are loaded **when they are needed**.

### Keras

Uses:

`tf.data.Dataset`

It can:

* Find images
* Infer labels
* Create batches
* Load data as needed

### PyTorch

Uses two parts:

**Dataset → gets one image/sample**

**DataLoader → creates batches and loads them efficiently**

```text
Dataset
   ↓
Individual images
   ↓
DataLoader
   ↓
Batches
   ↓
CNN
```

---

## 5. Data Augmentation

Data augmentation creates slightly different versions of training images.

Examples:

* 🔄 Rotate
* ↔️ Flip
* 🔍 Resize
* ✂️ Crop
* 🌈 Change appearance

### Important: On-the-fly augmentation

The augmented images are **not permanently saved**.

Instead:

```text
Original image
      ↓
Random transformation
      ↓
Training image
```

Every epoch can produce a slightly different version.

This helps the CNN **generalize better** and reduces overfitting.

### Keras

Uses:

* Preprocessing layers
* `ImageDataGenerator`

### PyTorch

Uses:

```python
torchvision.transforms
```

# 6. Building the CNN

### Keras

Keras can use the **Sequential API**:

```python
model = Sequential([
    Conv2D(...),
    MaxPooling2D(...),
    Flatten(),
    Dense(...)
])
```

Training is simple:

```python
model.fit(...)
```

Keras handles:

* Forward pass
* Backpropagation
* Optimizer updates
* Metrics

---

### PyTorch

In PyTorch, you usually create a class based on:

```python
nn.Module
```

You manually control the training process:

```text
Input
 ↓
Forward pass
 ↓
Calculate loss
 ↓
backward()
 ↓
Update weights
```

So:

**Keras → framework handles more for you**

**PyTorch → you control more yourself**

# 7. CNN Evaluation Metrics 📊

After training, we need to determine how well the model performs.

### Accuracy

Percentage of predictions that are correct.

$$
Accuracy = \frac{Correct\ Predictions}{Total\ Predictions}
$$

⚠️ Accuracy can be misleading when classes are highly imbalanced.

---

### Precision

> Of everything the model predicted as positive, how many were actually positive?

$$
Precision = \frac{TP}{TP+FP}
$$

Useful when **false positives are costly**.

---

### Recall

> Of all the actual positive examples, how many did the model correctly find?

$$
Recall = \frac{TP}{TP+FN}
$$

Useful when **missing a positive case is costly**.

---

### F1 Score

Combines precision and recall.

$$
F1 = 2 \times \frac{Precision \times Recall}
{Precision + Recall}
$$

It gives a balance between precision and recall.

---

### Confusion Matrix

Shows different types of predictions:

|                     | Predicted Positive | Predicted Negative |
| ------------------- | ------------------ | ------------------ |
| **Actual Positive** | TP                 | FN                 |
| **Actual Negative** | FP                 | TN                 |

It helps you see **which classes the model is confusing**.

---

### ROC & AUC

**ROC curve** shows the trade-off between:

* True Positive Rate
* False Positive Rate

**AUC** summarizes the ROC curve into a single value.

---

### IoU

**Intersection over Union**

Mostly used for:

* Object detection
* Image segmentation

It measures how much the predicted region overlaps with the actual region.

```text
IoU = Area of Overlap
      -----------------
      Area of Union
```

---

### mAP

**Mean Average Precision**

Commonly used for **object detection**.

It measures detection performance across different classes.

---

# 8. Important Keras vs PyTorch Difference ⚠️

When comparing Keras and PyTorch models, try to keep the following the same:

* Batch size
* Learning rate
* Optimizer
* Number of epochs
* Weight initialization
* Output configuration

### Softmax difference

This is particularly important.

**Keras** may put:

```python
Softmax
```

inside the model.

**PyTorch** commonly gives **raw logits** to the loss function, such as:

```python
nn.CrossEntropyLoss()
```

`CrossEntropyLoss` handles the appropriate softmax/log-softmax operation internally.
