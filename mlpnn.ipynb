````python
import torch
````

````python
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)

y_train_tensor = torch.tensor(y_train.to_numpy(), dtype=torch.float32)
y_test_tensor = torch.tensor(y_test.to_numpy(), dtype=torch.float32)
````

````python
from sklearn.preprocessing import StandardScaler

y_scaler = StandardScaler()

y_train_scaled = y_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

y_test_scaled = y_scaler.transform(
    y_test.to_numpy().reshape(-1, 1)
)
````

````python
y_train_tensor = torch.tensor(
    y_train_scaled,
    dtype=torch.float32
).squeeze()

y_test_tensor = torch.tensor(
    y_test_scaled,
    dtype=torch.float32
).squeeze()
````

````python
from torch.utils.data import Dataset

class HousePriceDataset(Dataset):
    def __init__(self, X,y):
        self.X=X
        self.y=y

    def __len__(self):
        return len(self.X)
    
    def __getitem__(self,idx):
        return self.X[idx], self.y[idx]
````

````python
train_dataset=HousePriceDataset(X_train_tensor,y_train_tensor)
test_dataset=HousePriceDataset(X_test_tensor,y_test_tensor)
````

````python
print(len(train_dataset))
print(train_dataset[0])
````

````python
train_dataset[0]
````

````python
from torch.utils.data import DataLoader

train_loader=DataLoader(train_dataset,batch_size=32,shuffle=True)
test_loader=DataLoader(test_dataset,batch_size=32,shuffle=False)
````

````python
print(len(train_loader))
````

````python
X_batch, y_batch = next(iter(train_loader))

print(X_batch.shape)
print(y_batch.shape)
````

````python
import torch
import torch.nn as nn

class HousePriceModel(nn.Module):

    def __init__(self, input_features):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_features, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 1)
        )

    def forward(self, x):
        return self.network(x)


model = HousePriceModel(25)

print(model)

predictions = model(X_batch)

print("Prediction shape:", predictions.shape)
````

Let's go through it line by line.

```python

```



Without ReLU:


With ReLU:


ReLU formula:

```python
ReLU(x) = max(0, x)
```

Examples:

```text
-5 → 0
 2 → 2
10 → 10
```

---

```python

```

Second hidden layer.

```text
64 neurons
   ↓
32 neurons
```

The model combines information learned by the first layer.

---

```python
nn.ReLU(),
```

Another activation function.

---

```python
nn.Linear(32, 1)
```

Output layer.

```text
32 neurons
   ↓
1 value
```

That one value is the predicted house price.

Example:

```text

```

```python
)
```


Defines how data moves through the network.

This single line does:

```python
x = nn.Linear(25, 64)(x)

return x
```

automatically.

---

### Complete Flow

If your batch shape is:

```python
X_batch.shape
```

Output:

```python
torch.Size([32, 25])
```

Then inside the model:

```text
[32, 25]
    ↓
Linear(25,64)
    ↓
[32, 64]
    ↓
ReLU
    ↓
[32, 64]
    ↓
Linear(64,32)
    ↓
[32, 32]
    ↓
ReLU
    ↓
[32, 32]
    ↓
Linear(32,1)
    ↓
[32, 1]
```

Final prediction shape:

```python
torch.Size([32, 1])
```

Meaning:

```text
House 1 → predicted price
House 2 → predicted price
House 3 → predicted price
...
House 32 → predicted price
```

This is exactly how a basic MLP (Multi-Layer Perceptron) regression model is built in PyTorch.


````python
loss_function = nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(),lr=0.001)
````

````python
predictions = model(X_batch)

loss = loss_function(predictions, y_batch.view(-1, 1))

print("Predictions:", predictions[:5])
print("Actual:", y_batch[:5])
print("Loss:", loss.item())
````

````python
epochs = 25
````

````python
train_losses = []
test_losses = []

for epoch in range(epochs):
    model.train()
    train_loss = 0.0

    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()

        predictions = model(X_batch)
        loss = loss_function(predictions, y_batch.view(-1, 1))

        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    train_loss /= len(train_loader)
    train_losses.append(train_loss)

    model.eval()
    test_loss = 0.0

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            predictions = model(X_batch)
            loss = loss_function(predictions, y_batch.view(-1, 1))
            test_loss += loss.item()

    test_loss /= len(test_loader)
    test_losses.append(test_loss)

    print(
        f"Epoch {epoch + 1}/{epochs} | "
        f"Train Loss: {train_loss:.3f} | "
        f"Test Loss: {test_loss:.3f}"
    )
````

````python
for epoch in range(epochs):
    model.train()
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        predictions = model(X_batch)
        loss = loss_function(predictions, y_batch.view(-1, 1))
        loss.backward()
        optimizer.step()

    model.eval()
    with torch.no_grad():
        test_loss = 0
        for X_batch, y_batch in test_loader:
            predictions = model(X_batch)
            loss = loss_function(predictions, y_batch.view(-1, 1))
            test_loss += loss.item()

    print(f"Epoch {epoch+1}, Test Loss: {test_loss/len(test_loader)}")
````

````python
model.eval()

predictions = []
actuals = []

with torch.no_grad():
    for X_batch, y_batch in test_loader:
        pred = model(X_batch)

        predictions.extend(pred.squeeze(1).numpy())
        actuals.extend(y_batch.numpy())
````

````python
predictions_original = y_scaler.inverse_transform(
    np.asarray(predictions).reshape(-1, 1)
).ravel()

actuals_original = y_scaler.inverse_transform(
    np.asarray(actuals).reshape(-1, 1)
).ravel()

predictions_original = y_scaler.inverse_transform(
    np.asarray(predictions).reshape(-1, 1)
).ravel()

actuals_original = y_scaler.inverse_transform(
    np.asarray(actuals).reshape(-1, 1)
).ravel()
````

````python
for pred, actual in zip(predictions_original[:10], actuals_original[:10]):
    print(f"Predicted: {pred:.0f} | Actual: {actual:.0f}")
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
mae = mean_absolute_error(actuals_original, predictions_original)

mse = mean_squared_error(actuals_original, predictions_original)

rmse = np.sqrt(mse)

r2 = r2_score(actuals_original, predictions_original)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
````
